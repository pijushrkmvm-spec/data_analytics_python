```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('diwali_sales_data.csv', encoding='unicode_escape')
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
df = df.drop(['Status', 'unnamed1'], axis=1)
# instead of saving it into df again we can use inplace=True parameter to save
```

```python
df.info()
```

```python
# checking null values
df.isnull().sum()
```

```python
# printing rows containing null values
null_rows = df[df.isnull().any(axis=1)]
print(null_rows)
```

```python
# null rows can be either removed or be filled with any value or mean, median or mode etc.
# removing null rows
# df = df.dropna()

# to fill with specific data (e.g. mean)
df['Amount'] = df['Amount'].fillna(df['Amount'].mean())
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
# the datatype of any column can be changed accordingly
df['Amount'] = df['Amount'].astype('int')
df.info()
```

```python
# column name can be changed using rename function
# df.rename(columns= {'Amount':'Money'})
```

```python
df['Amount'].min()
```

```python
df['Amount'].max()
```

```python
# we can selectively use describe fuction for specific columns
df[['Age','Orders','Amount']].describe()
```

### Exploratory data analysis

```python
df.columns
```

##### Analysis on Gender

```python
# generating a countplot using seaborn
sns.countplot(x = 'Gender', data = df, hue = 'Gender')
```

```python
# labeling each bars
ax = sns.countplot(data = df, x = "Gender", hue = "Gender")

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_per_gender = df.groupby(['Gender'], as_index = False)['Amount'].sum()
sns.barplot(x = 'Gender', y = 'Amount', data = sales_per_gender, hue = 'Gender')
```

##### Analysis on Age

```python
df.columns
```

```python
# age group that did most purchase
sns.countplot(data = df, x = 'Age Group')
```

```python
# division of gender per age group
sns.countplot(data = df, x = 'Age Group', hue = 'Gender')
```

```python
# sales per age group
sales_per_age_group = df.groupby(['Age Group'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)

sns.barplot(data = sales_per_age_group, x = 'Age Group', y = 'Amount', hue = 'Age Group')
```

##### Analysis on State

```python
df.columns
```

```python
sales_per_state = df.groupby(['State'], as_index = False)['Orders'].sum().sort_values(by = 'Orders', ascending = False)

sns.set(rc = {'figure.figsize': (20,5)})
sns.barplot(data = sales_per_state, x = 'State', y = 'Orders', hue = 'State')
```

```python
# for top 10 states
sales_per_state = df.groupby(['State'], as_index = False)['Orders'].sum().sort_values(by = 'Orders', ascending = False).head(10)

sns.set(rc = {'figure.figsize': (15,5)})
sns.barplot(data = sales_per_state, x = 'State', y = 'Orders', hue = 'State')
```

```python
# for top 10 states
sales_per_state = df.groupby(['State'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False).head(10)

sns.set(rc = {'figure.figsize': (15,5)})
sns.barplot(data = sales_per_state, x = 'State', y = 'Amount', hue = 'State')
```

##### Analysis on MaritalStatus

```python
df.columns
```

```python
sns.countplot(data = df, x = 'Marital_Status')
sns.set(rc = {'figure.figsize': (7,5)})
```

##### Analysis on Occupation

```python
sns.countplot(data = df, x = 'Occupation', hue = 'Occupation', order = df['Occupation'].value_counts().index)
sns.set(rc = {'figure.figsize': (20,5)})
```

```python
sales_per_occupation = df.groupby(['Occupation'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)

sns.set(rc = {'figure.figsize': (20,5)})
sns.barplot(data = sales_per_occupation, x = 'Occupation', y = 'Amount', hue = 'Occupation')
```

##### Analysis on ProductCategory

```python
df.columns
```

```python
sns.countplot(data = df, x = 'Product_Category', hue = 'Product_Category', order = df['Product_Category'].value_counts().head(10).index)
```

```python
sales_per_product = df.groupby(['Product_Category'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False).head(10)

sns.set(rc = {'figure.figsize': (20,5)})
sns.barplot(data = sales_per_product, x = 'Product_Category', y = 'Amount', hue = 'Product_Category')
```

```python
order_per_productID = df.groupby(['Product_ID'], as_index = False)['Orders'].sum().sort_values(by = 'Orders', ascending = False).head(10)

sns.barplot(data = order_per_productID, x = 'Product_ID', y = 'Orders', hue = 'Product_ID')
```
